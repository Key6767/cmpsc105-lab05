## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", palette="hsv")
ax.set_ylim(0, 5500) 
plt.title("Penguin Mass by Species")
plt.show()
```

Issues with the plot: The grams portion started at around 3000 grams, which obfuscates the true ammounts, showing 2 small ammounts in comparison to the bigger ones (Gentoo). 


How you changed the plot and why: Starting from 0 gives a more realistic image, where we can see that the 3 species (Adelie, and Chinstrap) are more than half the Gentoo in grams, instead of them being less than half (Which although less extreme, shows the reality of the masses captured in our data). 


## Plot 2

**Scenario**: A student plotted the carat size versus price for 54,000 diamonds to see where the bulk of the market sits, but the result is just a solid block of ink.

```python
diamonds = sns.load_dataset("diamonds")
plt.figure(figsize=(12, 6))
sns.lineplot(data=diamonds, x="carat", y="price", color="blue", linewidth=2) 
plt.xlim(0, 5)  
plt.ylim(0, 20000)
plt.title("Diamond Price vs. Carat")
plt.show()
```

Issues with the plot: Quite illegible with the dots, they fade away against each other because of the dot sizes. 


How you changed the plot and why: I changed the plot from scatter to line, in order to better communicate the trend of karat and price range. I also made the plot scale smaller, in order for the viewer to notice the trend rather than just seeing vertical lines, and I made the overall plot size bigger for better viewing. 

## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
flights = sns.load_dataset("flights")

sns.lineplot(data=flights, x="month", y="passengers", hue="year", palette="husl")
plt.title("Passengers over Time")
plt.show()
```

Issues with the plot: It's all one color


How you changed the plot and why: Made different colors, so one could tell which data belongs to which time. 


## Plot 4

**Scenario**: A restaurant manager receives this chart showing the relationship between total bills and tips, split by smoking preference. 





```python
tips = sns.load_dataset("tips")

sns.lmplot(data=tips, x="total_bill", y="tip", hue="smoker")
plt.title("Tip vs. Total Bill by Smoker Status") 
plt.show()
```

Issues with the plot: Does not have a title. 


How you changed the plot and why: Added a title, in order to identify what we are examining in this plot.
